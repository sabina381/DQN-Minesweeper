```python
from trainer import Trainer
from tester import Tester
from config import CONFIG
import pickle
from pathlib import Path
```

# Save parameter

```python
# 파라미터 파일 저장
path = f"{CONFIG.PATH}/{CONFIG.FOLDER_NAME}/config.pkl"
path = Path(path)
path.parent.mkdir(parents=True, exist_ok=True)

with open(path, "wb") as f:
    pickle.dump(CONFIG, f)

# 파라미터를 출력한 txt 파일 생성
path = f"{CONFIG.PATH}/{CONFIG.FOLDER_NAME}/config.txt"
path = Path(path)
path.parent.mkdir(parents=True, exist_ok=True)

if not path.exists():
    config_str = ""
    for item in CONFIG.items():
        config_str += str(item) + "\n"

    with open(path, "a") as f:
        f.write(config_str)
```

# Train

```python
trainer = Trainer(FOLDER_NAME=CONFIG.FOLDER_NAME,
            PATH=CONFIG.PATH,
            EPISODES=CONFIG.EPISODES,
            VALID_EPISODES=CONFIG.VALID_EPISODES,
            LAG=CONFIG.LAG,
            MODEL_CRITERIA=CONFIG.MODEL_CRITERIA,
            UPDATE_TARGET_EVERY=CONFIG.UPDATE_TARGET_EVERY,
            PRINT_EVERY=CONFIG.PRINT_EVERY,
            SAVE_EVERY=CONFIG.SAVE_EVERY,
            VALID_EVERY=CONFIG.VALID_EVERY,
            TRAIN_START=CONFIG.TRAIN_START,
            LEARN_MIN=CONFIG.LEARN_MIN,
            LEARN_DECAY=CONFIG.LEARN_DECAY,
            LEARN_EPOCH=CONFIG.LEARN_EPOCH,
            # Environment
            GRIDWORLD_SIZE=CONFIG.GRIDWORLD_SIZE,
            NUM_MINE=CONFIG.NUM_MINE,
            REWARD_DICT=CONFIG.REWARD_DICT,
            DONE_DICT=CONFIG.DONE_DICT,
            COLOR_DICT=CONFIG.COLOR_DICT,
            FIRST_MINE=CONFIG.FIRST_MINE,
            # DQN_Agent
            STATE_TYPE=CONFIG.STATE_TYPE,
            EPSILON=CONFIG.EPSILON,
            EPSILON_DECAY=CONFIG.EPSILON_DECAY,
            EPSILON_MIN=CONFIG.EPSILON_MIN,
            BATCH_SIZE=CONFIG.BATCH_SIZE,
            GAMMA=CONFIG.GAMMA,
            MEM_SIZE=CONFIG.MEM_SIZE,
            LEARN_MAX=CONFIG.LEARN_MAX,
            CONV_UNITS=CONFIG.CONV_UNITS,
            DEVICE=CONFIG.DEVICE,
            LOSS_FUNC=CONFIG.LOSS_FUNC,
            OPTIMIZER=CONFIG.OPTIMIZER)
```

```python
trainer.train()
```
